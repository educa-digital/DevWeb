### 🎯 Qué vas a aprender
Una matriz es una tabla rectangular de números. **NumPy** permite representarla y operar con ella.
- `import numpy as np` carga NumPy.
- `np.array([[...], [...]])` crea una matriz.
- `A.shape` devuelve filas y columnas.
- `A + B` suma elemento a elemento; `A * 3` multiplica por un escalar.
- `A @ B` realiza el producto matricial.
- `A.T` calcula la traspuesta.

Para el producto matricial, las columnas de la primera matriz deben coincidir con las filas de la segunda.


### 🔍 Ejemplo guiado · Crear y operar con matrices

Los índices de NumPy comienzan en cero: `A[0, 1]` accede a la primera fila y segunda columna. Observa las distintas operaciones y comprueba que `*` y `@` no son equivalentes.


```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 1],
              [2, 3]])

print("A:")
print(A)
print("Dimensiones:", A.shape)
print("Elemento fila 1, columna 2:", A[0, 1])
print("Suma A + B:\n", A + B)
print("Producto A por 2:\n", A * 2)
print("Producto matricial A @ B:\n", A @ B)
print("Traspuesta de A:\n", A.T)
print("Determinante de A:", np.linalg.det(A))
```

### 🎯 Qué vas a aprender
Un sistema lineal puede escribirse como \(A\cdot X=B\), donde \(A\) es la matriz de coeficientes, \(X\) el vector de incógnitas y \(B\) los términos independientes.

`np.linalg.solve(A, B)` calcula el vector solución. Para comprobarlo, calcula `A @ solucion` y compáralo con `B`.

El método requiere una matriz cuadrada invertible; si el determinante es cero, el sistema no tiene solución única.


### 🔍 Ejemplo guiado · Sistema de dos ecuaciones

Resolvemos \(2x+y=5\), \(x-y=1\). Cada fila de `A` contiene los coeficientes de una ecuación y `B` contiene sus términos independientes.


```python
import numpy as np

A = np.array([[2, 1],
              [1, -1]])
B = np.array([5, 1])

solucion = np.linalg.solve(A, B)
print("Solución [x, y]:", solucion)
print("x =", solucion[0])
print("y =", solucion[1])
print("Comprobación A @ solucion:", A @ solucion)
print("Términos independientes:", B)
```

### 🎯 Qué vas a aprender
**SymPy** permite realizar operaciones algebraicas simbólicas.
- `import sympy as sp` carga la biblioteca.
- `x = sp.symbols('x')` crea una variable simbólica.
- `sp.limit(expr, x, a)` calcula el límite cuando \(x\to a\).
- `expr.subs(x, valor)` sustituye la variable.
- `sp.simplify(expr)` simplifica una expresión.

Si la sustitución directa produce \(0/0\), existe una indeterminación y no podemos concluir el límite mediante esa sustitución.


### 🔍 Ejemplo guiado · Sustitución e indeterminación

Primero calculamos un límite directo. Después estudiamos una expresión que produce \(0/0\) al sustituir, y dejamos que SymPy calcule el límite simbólico.


```python
import sympy as sp

x = sp.symbols('x')
funcion = x**2 + 3*x
print("Límite cuando x tiende a 2:", sp.limit(funcion, x, 2))
print("Sustitución directa:", funcion.subs(x, 2))

funcion_2 = (x**2 - 4) / (x - 2)
print("Expresión:", funcion_2)
print("Sustitución directa:", funcion_2.subs(x, 2))
print("Límite:", sp.limit(funcion_2, x, 2))
print("Simplificación:", sp.simplify(funcion_2))
```

### 🎯 Qué vas a aprender
La derivada describe la tasa de variación instantánea. Geométricamente, \(f'(a)\) es la pendiente de la recta tangente en \(x=a\).

- `sp.diff(funcion, x)` deriva respecto de `x`.
- `expr.subs(x, a)` sustituye \(x\) por \(a\).
- `sp.lambdify(x, expr, "numpy")` convierte una expresión simbólica en función numérica.
- `np.linspace(inicio, fin, cantidad)` genera valores equidistantes.
- `plt.plot(x, y)` representa una curva.


### 🔍 Ejemplo guiado · Derivar y dibujar la tangente

Para \(f(x)=x^3-3x^2+2\), calculamos la derivada y su valor en 2. La tangente se construye con \(y=f(a)+f'(a)(x-a)\).


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
f = x**3 - 3*x**2 + 2
derivada = sp.diff(f, x)
a = 2
valor_f = f.subs(x, a)
pendiente = derivada.subs(x, a)
tangente = valor_f + pendiente*(x - a)

print("f'(x) =", derivada)
print("f'(2) =", pendiente)
print("Tangente:", sp.expand(tangente))

f_numpy = sp.lambdify(x, f, "numpy")
t_numpy = sp.lambdify(x, tangente, "numpy")
valores_x = np.linspace(-1, 4, 200)

plt.plot(valores_x, f_numpy(valores_x), label="f(x)")
plt.plot(valores_x, t_numpy(valores_x), label="Tangente en x=2")
plt.scatter([a], [float(valor_f)], label="Punto de tangencia")
plt.axhline(0); plt.axvline(0)
plt.grid(); plt.legend(); plt.show()
```

### 🎯 Qué vas a aprender
Los puntos críticos interiores se buscan resolviendo \(f'(x)=0\). Un punto crítico no es automáticamente un máximo o mínimo: hay que estudiar el signo de la derivada a ambos lados.

- `sp.solve(ecuacion, x)` resuelve una ecuación simbólica.
- `sp.factor(expr)` factoriza una expresión.
- Si \(f'(x)>0\), la función crece; si \(f'(x)<0\), decrece.
- `if`, `elif` y `else` permiten clasificar los resultados.


### 🔍 Ejemplo guiado · Puntos críticos y signo de la derivada

Para \(f(x)=x^3-3x\), calculamos la derivada, resolvemos \(f'(x)=0\) y probamos un valor en cada intervalo para conocer el signo de la derivada.


```python
import sympy as sp

x = sp.symbols('x')
f = x**3 - 3*x
derivada = sp.diff(f, x)
puntos = sp.solve(derivada, x)

print("Derivada:", derivada)
print("Puntos críticos:", puntos)

for punto in puntos:
    print("f(", punto, ") =", f.subs(x, punto))

for valor in [-2, 0, 2]:
    resultado = derivada.subs(x, valor)
    if resultado > 0:
        print("En x =", valor, "la función crece.")
    elif resultado < 0:
        print("En x =", valor, "la función decrece.")
    else:
        print("Derivada nula.")
```

### 🎯 Qué vas a aprender
Una primitiva \(F(x)\) de \(f(x)\) cumple \(F'(x)=f(x)\). La integral definida se calcula con la regla de Barrow: \(\int_a^b f(x)\,dx=F(b)-F(a)\).

- `sp.integrate(funcion, x)` calcula una primitiva.
- `sp.integrate(funcion, (x, a, b))` calcula una integral definida.
- `sp.diff()` permite comprobar la primitiva derivando el resultado.

SymPy devuelve una primitiva sin escribir la constante \(C\). La integral definida representa área con signo.


### 🔍 Ejemplo guiado · Integral indefinida y definida

Calculamos una primitiva de \(3x^2+2x\) y comprobamos derivándola. Después calculamos la integral definida de \(x^2\) entre 0 y 2.


```python
import sympy as sp

x = sp.symbols('x')
f = 3*x**2 + 2*x
primitiva = sp.integrate(f, x)

print("Primitiva:", primitiva)
print("Comprobación derivando:", sp.diff(primitiva, x))

funcion = x**2
integral_definida = sp.integrate(funcion, (x, 0, 2))
print("Integral entre 0 y 2:", integral_definida)
print("Valor decimal:", float(integral_definida))
```

- `sp.limit()` calcula límites.
- `sp.diff()` calcula derivadas.
- `sp.solve()` resuelve la ecuación de los puntos críticos.
- `sp.integrate()` calcula primitivas e integrales definidas.
- `sp.lambdify()` y Matplotlib permiten dibujar la función.


### 🔍 Ejemplo guiado · Informe matemático básico

Trabajaremos con \(f(x)=x^3-3x\). Observa cómo se agrupan límites, derivada, puntos críticos, primitiva e integral definida.


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
f = x**3 - 3*x

limite_mas = sp.limit(f, x, sp.oo)
limite_menos = sp.limit(f, x, -sp.oo)
derivada = sp.diff(f, x)
criticos = sp.solve(derivada, x)
primitiva = sp.integrate(f, x)
integral = sp.integrate(f, (x, 0, 2))

print("Función:", f)
print("Límite +infinito:", limite_mas)
print("Límite -infinito:", limite_menos)
print("Derivada:", derivada)
print("Puntos críticos:", criticos)
print("Primitiva:", primitiva)
print("Integral entre 0 y 2:", integral)
```

```python
# TU CÓDIGO — Sesión 8
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
g = x**3 - 6*x**2 + 9*x + 2

limite_mas = sp.limit(g, x, sp.oo)
limite_menos = sp.limit(g, x, -sp.oo)
derivada_g = sp.diff(g, x)
puntos_criticos = sp.solve(derivada_g, x)
primitiva_g = sp.integrate(g, x)
integral_g = sp.integrate(g, (x, 0, 3))

print("Función:", g)
print("Límite +infinito:", limite_mas)
print("Límite -infinito:", limite_menos)
print("Derivada:", derivada_g)
print("Puntos críticos:", puntos_criticos)
print("Primitiva:", primitiva_g)
print("Integral de 0 a 3:", integral_g)

g_numpy = sp.lambdify(x, g, "numpy")
valores_x = np.linspace(-1, 5, 300)
valores_y = g_numpy(valores_x)

plt.plot(valores_x, valores_y, label="g(x)")
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.legend()
plt.title("Estudio gráfico de g(x)")
plt.xlabel("x")
plt.ylabel("g(x)")
plt.show()
```

# **EXTRA 2: Plot interactivo con bokeh**
Con bokeh podemos crear plots interactivos que se sirven como una página web. Es muy útil cuando quieres mostrar datos a un usuario como gráficas, y permitir al usuario que interactúe con ellas.

A continuación vemos como implementar lo siguiente. Un diagrama con la representación de una onda sinusoidal y dos widgets que permiten modificar los parámetros de la misma.

```python
import numpy as np
from bokeh.io import curdoc
from bokeh.layouts import column, row
from bokeh.models import ColumnDataSource, Slider
from bokeh.plotting import figure
```

Definimos la onda sinusoidal. Con linspace podemos definir N puntos entre 0 y 2*PI.

```python
N = 200
x = np.linspace(0, 4*np.pi, N)
y = np.sin(x)
source = ColumnDataSource(data=dict(x=x, y=y))

plot = figure(height=400, width=400, y_range=[-2.5, 2.5])
plot.line('x', 'y', source=source, line_width=3, line_alpha=0.6)
```

Creamos dos Slider. Estos permitirán al usuario modificar la amplitud y frecuencia interactivamente. El on_change nos permite decir que hacer cuando la amplitud o frecuencia cambien.

```python
amplitud = Slider(title="amplitud", value=1.0, start=-5.0, end=5.0, step=0.1)
frecuencia = Slider(title="frecuencia", value=1.0, start=0.1, end=5.1, step=0.1)

def update_data(attrname, old, new):
    a = amplitud.value
    k = frecuencia.value

    x = np.linspace(0, 4*np.pi, N)
    y = a*np.sin(k*x)

    source.data = dict(x=x, y=y)

for w in [amplitud, frecuencia]:
    w.on_change('value', update_data)
```

Creamos el plot con los dos Slider y el plot.

```python
curdoc().add_root(
    column(amplitud,
           frecuencia,
           plot, width=800))
```

Y con el siguiente comando podemos ejecutar. El resultado se podrá ver en la siguiente dirección:

🔗 http://localhost:5006

bokeh serve interactivo_bokeh.py

```python
import time
import sys

def cuenta_atras(segundos):
    for queda in range(segundos, 0, -1):
        progreso = int((segundos - queda) / segundos * 30)
        barra = f"[{'=' * progreso}{' ' * (30 - progreso)}]"
        sys.stdout.write(f"\r{barra} {queda}s restantes...")
        sys.stdout.flush()
        time.sleep(1)
    print("\r[==============================] Completado!      ")

cuenta_atras(10)
```

```python
import random
import string

def generar_contraseña(longitud=12):
    # Caracteres disponibles
    caracteres = string.ascii_letters + string.digits + "!@#$%&*"

    # Generar contraseña aleatoria
    contraseña = ''.join(random.choice(caracteres) for _ in range(longitud))

    print(f"Tu nueva contraseña: {contraseña}")
    return contraseña

# Ejemplo de uso
generar_contraseña(16)  # Contraseña de 16 caracteres
```

```python
def calcular_imc(peso, altura):
    # Calcular IMC (peso en kg, altura en metros)
    imc = peso / (altura ** 2)

    # Clasificar según estándares OMS
    if imc < 18.5:
        categoria = "Bajo peso"
        emoji = "⚠️"
    elif 18.5 <= imc < 25:
        categoria = "Peso normal"
        emoji = "✅"
    elif 25 <= imc < 30:
        categoria = "Sobrepeso"
        emoji = "⚠️"
    else:
        categoria = "Obesidad"
        emoji = "🔴"

    print(f"IMC: {imc:.2f}")
    print(f"Categoría: {emoji} {categoria}")

    return imc, categoria

# Ejemplo de uso
calcular_imc(70, 1.75)  # 70 kg, 1.75 m
```

```python
import psutil
import time

def monitorear_sistema(intervalo=5):
    while True:
        # Obtener métricas
        cpu = psutil.cpu_percent(interval=1)
        memoria = psutil.virtual_memory().percent
        disco = psutil.disk_usage('/').percent

        # Alertas si supera umbrales
        alertas = []
        if cpu > 80: alertas.append(f"⚠️  CPU: {cpu}%")
        if memoria > 85: alertas.append(f"⚠️  Memoria: {memoria}%")
        if disco > 90: alertas.append(f"⚠️  Disco: {disco}%")

        # Mostrar estado
        print(f"\n--- Estado del Sistema ---")
        print(f"CPU: {cpu}% | RAM: {memoria}% | Disco: {disco}%")

        if alertas:
            print("\n🔴 ALERTAS:")
            for alerta in alertas:
                print(alerta)

        time.sleep(intervalo)

# Ejemplo de uso
monitorear_sistema()
```

```python
import shutil
from datetime import datetime
from pathlib import Path

def crear_backup(carpeta_origen, carpeta_destino):
    # Crear nombre con fecha actual
    fecha = datetime.now().strftime("%Y%m%d_%H%M%S")
    nombre_backup = f"backup_{fecha}"

    # Ruta completa del backup
    ruta_backup = Path(carpeta_destino) / nombre_backup

    # Crear copia comprimida
    shutil.make_archive(str(ruta_backup), 'zip', carpeta_origen)

    print(f"✓ Backup creado: {ruta_backup}.zip")
    return f"{ruta_backup}.zip"

# Ejemplo de uso
crear_backup("/ruta/documentos", "/ruta/backups")
```
