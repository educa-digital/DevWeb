Una **lista** guarda varios elementos en orden. Un **diccionario** relaciona una clave con un valor. Para un vocabulario, el lema puede ser la clave y su traducción el valor.

### 🧠 Nuevas herramientas
- `len(diccionario)` cuenta cuántas parejas clave-valor contiene.
- `diccionario.items()` permite recorrer cada clave y su valor.

### 🔍 Ejemplo guiado

Este ejemplo reúne las operaciones que utilizarás en la actividad:

1. Cuenta las entradas.
2. Recorre el diccionario y muestra cada lema con su traducción.
3. Usa `if` para destacar los lemas de más de cuatro caracteres.
4. Pide una palabra y busca su significado.


```python
vocabulario = {
    "aqua": "agua",
    "terra": "tierra",
    "vita": "vida",
    "philosophia": "amor a la sabiduría"
}

print("Número de términos:", len(vocabulario))

for lema, significado in vocabulario.items():
    print(lema, "→", significado)
    if len(lema) > 4:
        print("  Este lema tiene más de cuatro caracteres.")

consulta = input("Escribe un lema: ")
consulta = consulta.strip().lower()
print(vocabulario.get(consulta, "Palabra no encontrada"))
```

### 🏛️ Conexión con Latín y Griego
Un **bucle** repite instrucciones. `for` recorre los elementos de una lista, uno a uno. `zip()` permite recorrer dos listas a la vez, por ejemplo, personas y formas verbales.

### 🧠 Nuevas herramientas
- `enumerate(lista, start=1)` añade un número a cada elemento durante el recorrido.
- `in` permite comprobar si un texto contiene una secuencia: `"pl." in persona` será verdadero cuando la abreviatura aparezca.
- `if` y `else` permiten clasificar cada persona según el resultado de esa comprobación.


### 🔍 Ejemplo guiado

Observa cómo `zip()` relaciona cada persona con su forma verbal. `enumerate()` añade un número de orden. Después, `in` permite detectar si la abreviatura indica plural.

Ejecuta el ejemplo y comprueba qué ocurre al cambiar una persona o una forma.


```python
personas = ["1.ª sg.", "2.ª sg.", "3.ª sg.", "1.ª pl.", "2.ª pl.", "3.ª pl."]
formas = ["amo", "amas", "amat", "amamus", "amatis", "amant"]

for numero, (persona, forma) in enumerate(zip(personas, formas), start=1):
    if "pl." in persona:
        numero_gramatical = "plural"
    else:
        numero_gramatical = "singular"

    print(numero, persona, "→", forma, "(", numero_gramatical, ")")
```

### 🔍 Ejemplo guiado

El programa busca raíces dentro de una palabra y guarda las coincidencias en una lista. `append()` añade cada coincidencia. Al final, `if not coincidencias` comprueba si no se ha encontrado ninguna.

Prueba con `biologia`, `geografia` y `teatro`.


```python
raices = {
    "bio": "vida",
    "geo": "tierra",
    "filo": "amistad o afinidad",
    "aqua": "agua"
}

termino = "biologia"
coincidencias = []

for raiz, significado in raices.items():
    if raiz in termino:
        coincidencias.append(raiz)
        print(raiz, "→", significado)

if not coincidencias:
    print("No se han encontrado coincidencias.")
else:
    print("Coincidencias:", coincidencias)
```

```python
from collections import Counter
import matplotlib.pyplot as plt

texto = "Gallia est omnis divisa in partes tres. Gallia est omnis."

texto_limpio = texto.lower().replace(".", "").replace(",", "").replace(";", "")
palabras = texto_limpio.split()

ignorar = ["est", "in"]
palabras_filtradas = []

for palabra in palabras:
    if palabra not in ignorar:
        palabras_filtradas.append(palabra)

frecuencias = Counter(palabras_filtradas)
datos = frecuencias.most_common(5)

print("Frecuencias:", datos)

etiquetas = [dato[0] for dato in datos]
valores = [dato[1] for dato in datos]

plt.bar(etiquetas, valores)
plt.title("Frecuencia de palabras")
plt.xlabel("Palabra")
plt.ylabel("Apariciones")
plt.show()
```

### 🔍 Ejemplo guiado

Este ejemplo muestra el funcionamiento de un menú repetitivo. El programa sigue preguntando hasta que el usuario escribe `salir`. Si escribe `lista`, muestra el contenido del diccionario.

Ejecuta la celda y prueba las dos opciones. Pulsa `salir` para terminar.


```python
vocabulario = {
    "aqua": "agua",
    "terra": "tierra",
    "vita": "vida"
}

while True:
    consulta = input("Escribe un lema, 'lista' o 'salir': ").strip().lower()

    if consulta == "salir":
        print("Hasta pronto.")
        break
    elif consulta == "lista":
        for lema, significado in vocabulario.items():
            print(lema, "→", significado)
    else:
        print(vocabulario.get(consulta, "Palabra no encontrada"))
```
